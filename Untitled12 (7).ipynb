
السؤال 1 — قاموس المتغيرات
 <div dir="rtl" style="text-align: right; font-family: sans-serif; line-height: 1.6;">

بما إنه عندنا ~1500 متغير، فكّرت إنه التوثيق اليدوي الكامل مستحيل، فلاحظت إنّ أسماء الأعمدة نفسها فيها بادئة (w_ للأم، m_ للأب، E للـendline) ولاحقة (_bi, _who) تدلّ على النوع والموضوع. فقرّرت أربط كل متغير بقسمه آليًا من خلال اسمه، وأكتب معنى كل قسم مرة واحدة من ملف الـcodebook. اعتمدت على المعنى لا على نوع pandas، لأنّ pandas يضع أنماطًا غير صحيحة

```python
import pandas as pd, numpy as np, re, json

# 1) تحميل البيانات + التنظيف (فارغ/ثابت/نص شبه فارغ)
m = pd.read_csv(data_folder + '/womens_mens_mcm.tab', sep='\t', low_memory=False)
c = pd.read_csv(data_folder + '/child_level_mcm.tab',  sep='\t', low_memory=False)
def clean(df):
    drop=[col for col in df.columns if df[col].isna().all() or df[col].nunique(dropna=True)==1
          or (df[col].isna().mean()>0.85 and df[col].nunique()>150)]
    return df.drop(columns=drop)
m, c = clean(m), clean(c)

# 2) المعاني المفصّلة (من دليلك) — ارفعي var_meanings.json لـColab
MEAN = json.load(open(data_folder + '/var_meanings.json', encoding='utf-8'))

# 3) معنى القسم احتياطيًا لغير الموثّق تفصيليًا
SECTION_MEANING = {
 'تصميم الدراسة والمعرّفات':'معرّف الأسرة، الذراع التجريبي، الطبقة، القرية/المنطقة','إداري/حالة الاستبيان':'حالة الاستبيان والتأهّل',
 'الديموغرافيا':'العمر، التعليم، الحالة الزوجية، عدد الأطفال','التمكين واتخاذ القرار':'مَن يقرّر في نفقات ومجالات الأسرة',
 'العلاقة الزوجية وجودتها':'تحسّن/سوء العلاقة ونمط التواصل','الحمل وصحة الأم':'الرعاية قبل الولادة والرضاعة والولادة',
 'استرجاع الغذاء والتنوّع الغذائي':'هل أكل الطفل/الأم صنفًا خلال 24 ساعة','قياسات الطفل والتحصين':'وزن/طول/MUAC الطفل واللقاحات',
 'أمراض الطفل':'الإسهال والملاريا والسعال','معايير النوع والعنف الأسري':'مواقف المساواة والعنف الأسري',
 'الأصول والدخل والإنفاق':'الملكيات والدخل والإنفاق','تغذية راجعة عن الجلسات':'حضور جلسات HFI والتغييرات',
 'تنظيم الأسرة والخصوبة':'وسائل تنظيم الأسرة والخصوبة','الإصحاح (النظافة)':'المياه والمرحاض وغسل اليدين',
 'نمو الطفل':'مؤشرات التطوّر الحركي/اللغوي','معرفة الصحة والتغذية':'أسئلة معرفية عن التغذية',
 'مركز الضبط':'عزو النتائج للحظ أم للجهد','أخرى/غير مصنّفة':'بنود متفرّقة',
}
def baseN(col): return col[1:] if (col.startswith('E') and len(col)>1 and col[1] in 'wmcd') else col
def stem(col): return re.sub(r'^(w_|m_)','', baseN(col)).lower()
RULES = [
 (r'^(id|treat|control|wemp|whn|mhn|stratum|village|district|subcounty|sample|enum|date|time|consent|eligib)','تصميم الدراسة والمعرّفات'),
 (r'(status|survey|moved|wives|ineligib|^elig|away|labr|did_wo|did_men|dum_|^a_)','إداري/حالة الاستبيان'),
 (r'(age|educ|relationship|marital|numwives|orgs|numadult|biokids|numkids|hhkids)','الديموغرافيا'),
 (r'(diffop|final|_say|rights|wmoney|chealth|cfeed|daily|wcloth|discuss|express|husknow|priority|dec|_who|opn|arg|prev)','التمكين واتخاذ القرار'),
 (r'(rel_improve|rel_worse|^rel_|improve|worse|commsty|satmon|huslstn)','العلاقة الزوجية وجودتها'),
 (r'(assist|comp|hus_ac|hus_preg|^ac|_ac|preg|malaria|breastf|colostr|birth|deliver|antenat)','الحمل وصحة الأم'),
 (r'(milk|meat|egg|fish|matooke|rice|beans|cas|organ|green|pump|fruit|dairy|oil|sweet|solidfood|tea|water|juice|food|porr|roots|starch|protein)','استرجاع الغذاء والتنوّع الغذائي'),
 (r'(weight|height|immcard|polio|dpt|measles|vita|iron|bcg|anthro|gender|muac|bmi|flag|child|feed|vacc|hblevel)','قياسات الطفل والتحصين'),
 (r'(sick|diarr|_mal|cough|worms|bednet|fever)','أمراض الطفل'),
 (r'(rght|equal|obey|boss|lead|choose|powhh|shrpow|opin|just|beat|slap|push|vio|harm|hum|unfaith|femfri|insts|notrust|listen|angry|quar|alch|hit_)','معايير النوع والعنف الأسري'),
 (r'(inc|emp|occup|wge|earn|save|ugx|spend|sug|soap|salt|transp|air|asset|own|hlth|schfee|womclt|supp|allow|disclose|tobacco|snacks|elec)','الأصول والدخل والإنفاق'),
 (r'(sessn|attend|hfi|getinfo|complexity|pace|notes|hchange|wechange|talkw|hushappy|healthinfo|weinfo)','تغذية راجعة عن الجلسات'),
 (r'(^fp|_fp|hiv|sex|contracep|condom|num_kids|ideal|fert|nextpreg)','تنظيم الأسرة والخصوبة'),
 (r'(toilet|latrine|handwash|sweep|refuse|hyg|premeal|postbath|wash|dispose|treat_water)','الإصحاح (النظافة)'),
 (r'(sound|smile|grasp|crawl|walk|word|jump|run|stack|play)','نمو الطفل'),
 (r'(phc|kwas|lackdiet|witch|deworm|anem|foodplate|menprot|menfruit|aniprot|cowmilk|cause|fteeth|barrier|true|false)','معرفة الصحة والتغذية'),
 (r'(luck|heredity|success|locus)','مركز الضبط'),
]
def section(col):
    st = stem(col)
    for pat,name in RULES:
        if re.search(pat, st): return name
    return 'أخرى/غير مصنّفة'

def clean_name(n): return re.sub(r'\s*\((child|BL|Baseline|EL|Endline)\)\s*','',n,flags=re.I).strip()
def vinfo(col, df):
    s=df[col]; nona=s.dropna(); vals=set(nona.unique())
    if col.endswith('_bi') or vals<={0,1} or vals<={0.0,1.0}: return 'ثنائي','{0 ، 1}'
    if col.endswith('_who'): return 'اسمي','1=زوجة 2=زوج 3=الاثنان 4=آخر 5=خارج'
    if s.dtype==object: return 'اسمي/نص', ' | '.join(map(str,list(vals)[:4]))
    if pd.api.types.is_numeric_dtype(s):
        a,b=nona.min(),nona.max()
        fa=str(int(a)) if a==int(a) else f'{a:.2f}'; fb=str(int(b)) if b==int(b) else f'{b:.2f}'
        return 'رقمي', f'[{fa} .. {fb}]'
    return 'غير محدد','-'

# 4) الدمج على كل المتغيرات (الأهل + الطفل)
seen=set(); rows=[]
for df in [m, c]:
    for col in df.columns:
        if col in seen: continue
        seen.add(col)
        meaning = MEAN.get(col) or MEAN.get(clean_name(col)) or SECTION_MEANING[section(col)]
        t,rng = vinfo(col, df)
        rows.append({'اسم المتغير':col,'المعنى':meaning,'النوع':t,'مجال القيم':rng})

data_dictionary = pd.DataFrame(rows)
detailed = data_dictionary['المعنى'].isin(MEAN.values()).sum()
print('إجمالي الموثّق:', len(data_dictionary), '| بمعنى مفصّل:', detailed, '| بمعنى قسم:', len(data_dictionary)-detailed)
data_dictionary.head(40)
```

```python
# 1) عرض المفقود
(c.isna().mean()*100).sort_values(ascending=False).head(15)

# 2) مقنّع: عمر الطفل المستحيل (>216 شهر = 18 سنة)
print('قيم عمر مستحيلة:', (c['age'] > 216).sum())
c.loc[c['age'] > 216, 'age'] = np.nan       # نحوّلها NaN

# 3) معالجة: c_weight مفقود بالتصميم → ما منعوّض، منستثني عند التحليل
c_anthro = c.dropna(subset=['c_weight'])     # العينة المقاسة فقط
```

س8 — القيم الشاذة + تبرير الطريقة

```python
def iqr_outliers(s):
    q1, q3 = s.quantile(.25), s.quantile(.75)
    iqr = q3 - q1
    lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
    return s[(s < lo) | (s > hi)]

for col, df_ in [('c_weight',c_anthro), ('Ec_weight',c), ('w_age',m), ('m_age',m)]:
    out = iqr_outliers(df_[col].dropna())
    print(f'{col}: {len(out)} شاذة ({len(out)/df_[col].notna().sum()*100:.1f}%)')
px.box(c_anthro, y='c_weight', points='all').show()
```

س9 — هل كانت التجربة عشوائية فعلًا؟

```python
from scipy.stats import f_oneway
m['group'] = m['treat'].map({1:'control',2:'wemp',3:'whn',4:'mhn'})
print(m.groupby('group')[['w_age','m_age','w_educ','m_educ','w_inc','m_inc']].mean())

for col in ['w_age','m_age','w_educ','m_educ']:
    groups = [g[col].dropna() for _, g in m.groupby('group')]
    f, p = f_oneway(*groups)
    print(f'{col}: p = {p:.3f}', '→ متوازن ✓' if p > 0.05 else '→ غير متوازن ✗')
```

س10 — أوزان الأطفال واستهلاك المنتجات الحيوانية قبل/بعد + أي تدريب أفضل؟

```python
from scipy.stats import ttest_rel, f_oneway
c['group'] = c['treat'].map({1:'control',2:'wemp',3:'whn',4:'mhn'})
c['dw'] = c['Ec_weight'] - c['c_weight']            # تغيّر الوزن
c['danim'] = c[['Emeat','Eegg','Efish','Emilk']].sum(axis=1) - c[['meat','egg','fish','milk']].sum(axis=1)

print(c.groupby('group')[['dw','danim']].mean())     # أي مجموعة الأعلى؟

# paired test داخل wemp: هل الوزن تحسّن فعلاً؟
w = c[c['group']=='wemp'].dropna(subset=['c_weight','Ec_weight'])
t, p = ttest_rel(w['Ec_weight'], w['c_weight'])
print(f'wemp قبل/بعد: t={t:.2f}, p={p:.4f}')

# ANOVA: هل الفرق بين المجموعات دال؟
groups = [g['dw'].dropna() for _, g in c.groupby('group')]
print('ANOVA على تغيّر الوزن: p =', f_oneway(*groups).pvalue)
```

```python
from scipy.stats import spearmanr
d = c.dropna(subset=['Ec_weight','Esick_days'])
rho, p = spearmanr(d['Ec_weight'], d['Esick_days'])
print(f'Spearman: rho={rho:.3f}, p={p:.4f}')
px.scatter(d, x='Esick_days', y='Ec_weight', trendline='ols').show()
```

س14 — فروقات عمرية زوج/زوجة + من يقرر صحة الطفل

```python
from scipy.stats import ttest_1samp, f_oneway
m['age_gap'] = m['m_age'] - m['w_age']
print('متوسط فرق العمر:', m['age_gap'].mean())
print('اختبار ≠0:', ttest_1samp(m['age_gap'].dropna(), 0))

sub = m.dropna(subset=['age_gap','w_chealth_who'])
print(sub.groupby('w_chealth_who')['age_gap'].mean())
groups = [g['age_gap'] for _, g in sub.groupby('w_chealth_who')]
print('ANOVA: p =', f_oneway(*groups).pvalue)
```
