```python
# ============================================================
# إعادة تحميل البيانات بـ encoding صحيح
# ============================================================

import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

data_folder = '/content/drive/My Drive/DM Homework 2 datasets'

# حمّل البيانات بـ low_memory=False لتجنب التحذيرات
df_main = pd.read_csv(
    os.path.join(data_folder, 'womens_mens_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

df_child = pd.read_csv(
    os.path.join(data_folder, 'child_level_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

print("✅ تم التحميل بنجاح!")
print(f"womens_mens_mcm: {len(df_main)} صف، {len(df_main.columns)} عمود")
print(f"child_level_mcm: {len(df_child)} صف، {len(df_child.columns)} عمود")

# ============================================================
# الطلب الأول: قاموس المتغيرات المختصر والمركز
# ============================================================

print("\n" + "=" * 100)
print("📋 الطلب الأول: قاموس المتغيرات")
print("=" * 100)

# دالة لإنشاء القاموس
def create_variables_dict(df, name="Dataset"):
    """
    إنشاء قاموس شامل لكل متغيرات الـ DataFrame
    """
    variables_dict = {}

    for column in df.columns:
        data_type = df[column].dtype

        # تصنيف نوع المتغير
        if data_type == 'object':
            variable_type = 'Nominal (Categorical)'
            unique_vals = df[column].unique()
            if len(unique_vals) <= 8:
                value_range = str(unique_vals[:8])
            else:
                value_range = f"({len(unique_vals)} قيم مختلفة)"

        elif data_type in ['int64', 'int32']:
            min_val = df[column].min()
            max_val = df[column].max()

            # تحديد هل هو ثنائي أم لا
            unique_vals = df[column].unique()
            if len(unique_vals) == 2:
                variable_type = 'Binary'
            else:
                variable_type = 'Numeric (Discrete)'

            value_range = f"[{min_val} to {max_val}]"

        elif data_type in ['float64', 'float32']:
            min_val = df[column].min()
            max_val = df[column].max()
            unique_vals = df[column].unique()

            # تحديد هل هو ثنائي (0.0, 1.0)
            if len(unique_vals) == 2 and set(unique_vals) == {0.0, 1.0}:
                variable_type = 'Binary'
            else:
                variable_type = 'Numeric (Continuous)'

            value_range = f"[{min_val:.2f} to {max_val:.2f}]"

        else:
            variable_type = str(data_type)
            value_range = 'Unknown'

        # حساب الإحصائيات
        missing_count = df[column].isna().sum()
        non_null_count = len(df) - missing_count
        missing_pct = (missing_count / len(df)) * 100
        unique_count = df[column].nunique()

        # مثال من البيانات
        if non_null_count > 0:
            example = str(df[column].dropna().iloc[0])[:50]
        else:
            example = 'All NaN'

        variables_dict[column] = {
            'Variable Name': column,
            'Data Type (Python)': str(data_type),
            'Variable Type': variable_type,
            'Value Range': value_range,
            'Unique Values': unique_count,
            'Non-Missing Count': non_null_count,
            'Missing Count': missing_count,
            'Missing %': f"{missing_pct:.1f}%",
            'Example': example
        }

    return variables_dict

# إنشاء القاموس للملفين
print("\n" + "-" * 100)
print("📊 womens_mens_mcm.tab المتغيرات")
print("-" * 100)

vars_main = create_variables_dict(df_main, "Women & Men")
vars_main_df = pd.DataFrame.from_dict(vars_main, orient='index')

# عرض أول 15 متغير
print("\nأول 15 متغير:")
print(vars_main_df.head(15).to_string())

print(f"\n... وإجمالي {len(vars_main)} متغير في هذا الملف")

print("\n" + "-" * 100)
print("📊 child_level_mcm.tab المتغيرات")
print("-" * 100)

vars_child = create_variables_dict(df_child, "Child Level")
vars_child_df = pd.DataFrame.from_dict(vars_child, orient='index')

print("\nجميع المتغيرات:")
print(vars_child_df.to_string())

# ============================================================
# ملخص إحصائي سريع
# ============================================================

print("\n" + "=" * 100)
print("📈 ملخص الإحصائيات الأساسية")
print("=" * 100)

print("\n1️⃣ womens_mens_mcm.tab:")
print(f"   • إجمالي الصفوف (الأزواج): {len(df_main)}")
print(f"   • إجمالي الأعمدة: {len(df_main.columns)}")
print(f"   • تصنيف المتغيرات:")

main_categorical = sum(1 for v in vars_main.values() if 'Nominal' in v['Variable Type'])
main_numeric = sum(1 for v in vars_main.values() if 'Numeric' in v['Variable Type'] and 'Binary' not in v['Variable Type'])
main_binary = sum(1 for v in vars_main.values() if 'Binary' in v['Variable Type'])

print(f"     - اسمية (Nominal): {main_categorical}")
print(f"     - رقمية (Numeric): {main_numeric}")
print(f"     - ثنائية (Binary): {main_binary}")

print("\n2️⃣ child_level_mcm.tab:")
print(f"   • إجمالي الصفوف (الأطفال): {len(df_child)}")
print(f"   • إجمالي الأعمدة: {len(df_child.columns)}")
print(f"   • تصنيف المتغيرات:")

child_categorical = sum(1 for v in vars_child.values() if 'Nominal' in v['Variable Type'])
child_numeric = sum(1 for v in vars_child.values() if 'Numeric' in v['Variable Type'] and 'Binary' not in v['Variable Type'])
child_binary = sum(1 for v in vars_child.values() if 'Binary' in v['Variable Type'])

print(f"     - اسمية (Nominal): {child_categorical}")
print(f"     - رقمية (Numeric): {child_numeric}")
print(f"     - ثنائية (Binary): {child_binary}")
```

```python

# ============================================================
# الطلب الثاني: عدد الأزواج والأطفال
# ============================================================

print("\n" + "=" * 100)
print("👨‍👩‍👧 الطلب الثاني: إحصائيات الأسر والأطفال")
print("=" * 100)

num_pairs = len(df_main)
num_children = len(df_child)
avg_children = num_children / num_pairs

print(f"\n✅ عدد الأزواج المشاركة: {num_pairs} زوج/زوجة")
print(f"✅ عدد الأطفال المدروسين: {num_children} طفل")
print(f"✅ متوسط عدد الأطفال لكل عائلة: {avg_children:.2f} طفل/عائلة")

# معلومات إضافية عن المجموعات
print("\n📊 توزيع المجموعات:")

if 'treat' in df_main.columns:
    treat_counts = df_main['treat'].value_counts().sort_index()
    print(f"\nتوزيع المجموعات (treat):")
    for treat_val, count in treat_counts.items():
        print(f"   • المجموعة {int(treat_val)}: {count} عائلة")

if 'wemp' in df_main.columns:
    wemp_counts = df_main['wemp'].value_counts()
    print(f"\nالأمهات المستفيدات من التدريب (wemp):")
    print(f"   • نعم: {wemp_counts.get(1, 0)} أم")
    print(f"   • لا: {wemp_counts.get(0, 0)} أم")

# ============================================================
# حفظ النتائج
# ============================================================

print("\n" + "=" * 100)
print("💾 حفظ النتائج")
print("=" * 100)

output_file = '/content/drive/My Drive/Task1_Complete_Dictionary.xlsx'

try:
    with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
        # ورقة 1: ملخص المتغيرات - womens_mens
        vars_main_df.to_excel(writer, sheet_name='WM Variables', index=True)

        # ورقة 2: ملخص المتغيرات - child_level
        vars_child_df.to_excel(writer, sheet_name='Child Variables', index=True)

        # ورقة 3: عينة من البيانات
        df_main.head(30).to_excel(writer, sheet_name='WM Sample Data', index=False)

        # ورقة 4: عينة من بيانات الأطفال
        df_child.head(30).to_excel(writer, sheet_name='Child Sample Data', index=False)

        # ورقة 5: الملخص الإحصائي
        summary_data = {
            'Metric': [
                'عدد الأزواج',
                'عدد الأطفال',
                'متوسط أطفال/عائلة',
                'عدد الأعمدة (WM)',
                'عدد الأعمدة (Child)',
                'متغيرات اسمية (WM)',
                'متغيرات رقمية (WM)',
                'متغيرات ثنائية (WM)',
                'متغيرات اسمية (Child)',
                'متغيرات رقمية (Child)',
                'متغيرات ثنائية (Child)'
            ],
            'Value': [
                num_pairs,
                num_children,
                f"{avg_children:.2f}",
                len(df_main.columns),
                len(df_child.columns),
                main_categorical,
                main_numeric,
                main_binary,
                child_categorical,
                child_numeric,
                child_binary
            ]
        }
        summary_df = pd.DataFrame(summary_data)
        summary_df.to_excel(writer, sheet_name='Summary', index=False)

    print(f"✅ تم الحفظ بنجاح:")
    print(f"   {output_file}")

except Exception as e:
    print(f"⚠️ خطأ في الحفظ: {e}")

print("\n" + "=" * 100)
print("✅ انتهى الطلب الأول والثاني")
print("=" * 100)
```

```python
# ============================================================
# الطلب الثالث: إحصائيات وتحليل المتغيرات الاسمية
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

# تعريب الخطوط في الرسوم البيانية
plt.rcParams['font.family'] = 'DejaVu Sans'
sns.set_style("whitegrid")

data_folder = '/content/drive/My Drive/DM Homework 2 datasets'

# تحميل البيانات
df_main = pd.read_csv(
    os.path.join(data_folder, 'womens_mens_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

df_child = pd.read_csv(
    os.path.join(data_folder, 'child_level_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

print("=" * 100)
print("الطلب الثالث والرابع: تحليل المتغيرات الاسمية (Nominal)")
print("=" * 100)

# ============================================================
# تحديد المتغيرات الاسمية
# ============================================================

def identify_nominal_vars(df):
    """
    تحديد جميع المتغيرات الاسمية (Categorical/Text)
    """
    nominal_vars = []
    for col in df.columns:
        if df[col].dtype == 'object':  # نصي
            nominal_vars.append(col)
    return nominal_vars

nominal_main = identify_nominal_vars(df_main)
nominal_child = identify_nominal_vars(df_child)

print(f"\n📋 المتغيرات الاسمية في womens_mens_mcm:")
print(f"   عدد المتغيرات: {len(nominal_main)}")
print(f"   أسماؤها: {nominal_main}")

print(f"\n📋 المتغيرات الاسمية في child_level_mcm:")
print(f"   عدد المتغيرات: {len(nominal_child)}")
print(f"   أسماؤها: {nominal_child}")

# ============================================================
# الطلب الثالث: إحصائيات وجداول التكرار لكل متغير اسمي
# ============================================================

print("\n" + "=" * 100)
print("الطلب الثالث: الإحصائيات والنتائج المثيرة للاهتمام")
print("=" * 100)

# دالة لحساب الإحصائيات
def analyze_nominal_var(df, var_name, data_name="Dataset"):
    """
    تحليل شامل لمتغير اسمي واحد
    """
    print(f"\n{'─' * 100}")
    print(f"📊 المتغير: {var_name}")
    print(f"{'─' * 100}")

    # حساب التكرارات
    value_counts = df[var_name].value_counts(dropna=False)
    total = len(df)

    print(f"\nإجمالي الحالات: {total}")
    print(f"عدد القيم المختلفة: {df[var_name].nunique()}")
    print(f"القيم المفقودة (NaN): {df[var_name].isna().sum()}")

    # جدول مفصل
    print(f"\n📈 توزيع التكرارات:")
    print(f"{'─' * 100}")
    print(f"{'القيمة':<30} {'التكرار':<15} {'النسبة %':<15}")
    print(f"{'─' * 100}")

    for value, count in value_counts.items():
        percentage = (count / total) * 100
        value_str = str(value)[:30] if pd.notna(value) else "Missing (NaN)"
        print(f"{value_str:<30} {count:<15} {percentage:>6.2f}%")

    # النتائج المثيرة للاهتمام
    print(f"\n💡 النتائج المثيرة للاهتمام:")

    # 1. القيمة الأكثر تكراراً
    most_common = value_counts.idxmax()
    most_common_count = value_counts.max()
    most_common_pct = (most_common_count / total) * 100
    print(f"   • القيمة الأكثر تكراراً: {most_common} ({most_common_count} حالة، {most_common_pct:.1f}%)")

    # 2. القيمة الأقل تكراراً (غير المفقودة)
    non_nan_counts = df[var_name].value_counts(dropna=True)
    if len(non_nan_counts) > 1:
        least_common = non_nan_counts.idxmin()
        least_common_count = non_nan_counts.min()
        least_common_pct = (least_common_count / total) * 100
        print(f"   • القيمة الأقل تكراراً: {least_common} ({least_common_count} حالة، {least_common_pct:.1f}%)")

    # 3. نسبة المفقودة
    missing_count = df[var_name].isna().sum()
    missing_pct = (missing_count / total) * 100
    if missing_pct > 0:
        print(f"   • نسبة البيانات المفقودة: {missing_pct:.1f}% ({missing_count} حالة)")
    else:
        print(f"   • لا توجد بيانات مفقودة ✅")

    # 4. التنوع
    unique_non_nan = df[var_name].nunique()
    print(f"   • عدد الفئات المختلفة: {unique_non_nan}")

    return value_counts

# ============================================================
# تحليل المتغيرات الاسمية من womens_mens_mcm
# ============================================================

print("\n\n" + "🔷" * 50)
print("البيانات: womens_mens_mcm.tab")
print("🔷" * 50)

nominal_analysis_main = {}

for var in nominal_main:
    nominal_analysis_main[var] = analyze_nominal_var(df_main, var, "womens_mens")

# ============================================================
# تحليل المتغيرات الاسمية من child_level_mcm
# ============================================================

print("\n\n" + "🔷" * 50)
print("البيانات: child_level_mcm.tab")
print("🔷" * 50)

nominal_analysis_child = {}

for var in nominal_child:
    nominal_analysis_child[var] = analyze_nominal_var(df_child, var, "child_level")

# ============================================================
# الطلب الرابع: الرسوم البيانية والجداول
# ============================================================

print("\n\n" + "=" * 100)
print("الطلب الرابع: الرسوم البيانية (Bar Charts)")
print("=" * 100)

# دالة لرسم Bar Chart
def plot_nominal_var(df, var_name, data_name="Dataset"):
    """
    رسم bar chart لمتغير اسمي
    """
    value_counts = df[var_name].value_counts(dropna=True).head(15)  # أعلى 15 فئة

    fig, ax = plt.subplots(figsize=(12, 6))

    bars = ax.bar(range(len(value_counts)), value_counts.values, color='steelblue', alpha=0.7)
    ax.set_xticks(range(len(value_counts)))
    ax.set_xticklabels([str(v)[:20] for v in value_counts.index], rotation=45, ha='right')

    ax.set_xlabel(f'القيم ({var_name})', fontsize=12, fontweight='bold')
    ax.set_ylabel('التكرار (عدد الحالات)', fontsize=12, fontweight='bold')
    ax.set_title(f'توزيع المتغير: {var_name}\n({data_name})', fontsize=14, fontweight='bold')

    # إضافة القيم على أعمدة الرسم
    for i, (bar, val) in enumerate(zip(bars, value_counts.values)):
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 20,
                str(val), ha='center', va='bottom', fontweight='bold')

    ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()

    return fig

# رسم المتغيرات الاسمية من womens_mens
print("\n📊 رسم المتغيرات الاسمية من womens_mens_mcm:")

for i, var in enumerate(nominal_main, 1):
    print(f"   {i}. {var}")
    fig = plot_nominal_var(df_main, var, "womens_mens_mcm")
    plt.savefig(f'/content/drive/My Drive/WM_Nominal_{i}_{var}.png', dpi=150, bbox_inches='tight')
    plt.close()

print(f"✅ تم رسم {len(nominal_main)} رسم بياني")

# رسم المتغيرات الاسمية من child_level
print("\n📊 رسم المتغيرات الاسمية من child_level_mcm:")

for i, var in enumerate(nominal_child, 1):
    print(f"   {i}. {var}")
    fig = plot_nominal_var(df_child, var, "child_level_mcm")
    plt.savefig(f'/content/drive/My Drive/Child_Nominal_{i}_{var}.png', dpi=150, bbox_inches='tight')
    plt.close()

print(f"✅ تم رسم {len(nominal_child)} رسم بياني")

# ============================================================
# Pie Charts إضافية
# ============================================================

print("\n📊 رسم Pie Charts للمتغيرات الرئيسية:")

# اختيار أهم المتغيرات الاسمية
important_vars_main = ['district_name', 'stratum'] if 'district_name' in nominal_main else nominal_main[:2]
important_vars_child = [v for v in nominal_child if df_child[v].nunique() <= 15][:3]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('توزيع المتغيرات الاسمية الرئيسية', fontsize=16, fontweight='bold')

# Pie charts من womens_mens
for idx, var in enumerate(important_vars_main[:2]):
    ax = axes[0, idx]
    value_counts = df_main[var].value_counts(dropna=True).head(10)
    ax.pie(value_counts.values, labels=[str(v)[:15] for v in value_counts.index],
           autopct='%1.1f%%', startangle=90)
    ax.set_title(f'womens_mens: {var}', fontweight='bold')

# Pie charts من child_level
for idx, var in enumerate(important_vars_child[:2]):
    ax = axes[1, idx]
    value_counts = df_child[var].value_counts(dropna=True).head(10)
    ax.pie(value_counts.values, labels=[str(v)[:15] for v in value_counts.index],
           autopct='%1.1f%%', startangle=90)
    ax.set_title(f'child_level: {var}', fontweight='bold')

plt.tight_layout()
plt.savefig('/content/drive/My Drive/Nominal_Pie_Charts.png', dpi=150, bbox_inches='tight')
plt.close()

print("✅ تم رسم Pie Charts")

# ============================================================
# جداول ملخصة في Excel
# ============================================================

print("\n💾 حفظ جداول التكرار في Excel:")

output_file = '/content/drive/My Drive/Task3_4_Nominal_Analysis.xlsx'

try:
    with pd.ExcelWriter(output_file, engine='openpyxl') as writer:

        # ورقة 1: ملخص womens_mens
        summary_rows = []
        for var in nominal_main:
            vc = df_main[var].value_counts(dropna=False)
            for value, count in vc.items():
                pct = (count / len(df_main)) * 100
                summary_rows.append({
                    'Variable': var,
                    'Value': str(value),
                    'Count': count,
                    'Percentage': f"{pct:.2f}%"
                })

        summary_df_main = pd.DataFrame(summary_rows)
        summary_df_main.to_excel(writer, sheet_name='WM Nominal Summary', index=False)

        # ورقة 2: ملخص child_level
        summary_rows = []
        for var in nominal_child:
            vc = df_child[var].value_counts(dropna=False)
            for value, count in vc.items():
                pct = (count / len(df_child)) * 100
                summary_rows.append({
                    'Variable': var,
                    'Value': str(value),
                    'Count': count,
                    'Percentage': f"{pct:.2f}%"
                })

        summary_df_child = pd.DataFrame(summary_rows)
        summary_df_child.to_excel(writer, sheet_name='Child Nominal Summary', index=False)

        # ورقات منفصلة لكل متغير (womens_mens)
        for var in nominal_main:
            vc = df_main[var].value_counts(dropna=False)
            vc_df = pd.DataFrame({
                'Value': vc.index,
                'Count': vc.values,
                'Percentage': [f"{(c/len(df_main))*100:.2f}%" for c in vc.values]
            })
            sheet_name = f"WM_{var[:20]}"
            vc_df.to_excel(writer, sheet_name=sheet_name, index=False)

        # ورقات منفصلة لكل متغير (child_level)
        for var in nominal_child:
            vc = df_child[var].value_counts(dropna=False)
            vc_df = pd.DataFrame({
                'Value': vc.index,
                'Count': vc.values,
                'Percentage': [f"{(c/len(df_child))*100:.2f}%" for c in vc.values]
            })
            sheet_name = f"Child_{var[:15]}"
            vc_df.to_excel(writer, sheet_name=sheet_name, index=False)

    print(f"✅ تم الحفظ: {output_file}")

except Exception as e:
    print(f"⚠️ خطأ: {e}")

# ============================================================
# ملخص الاستنتاجات
# ============================================================

print("\n" + "=" * 100)
print("📌 الاستنتاجات الرئيسية")
print("=" * 100)

print("\n🔍 womens_mens_mcm.tab:")
print(f"   • عدد المتغيرات الاسمية: {len(nominal_main)}")
print(f"   • أسماء المتغيرات: {', '.join(nominal_main)}")

print("\n🔍 child_level_mcm.tab:")
print(f"   • عدد المتغيرات الاسمية: {len(nominal_child)}")
print(f"   • أسماء المتغيرات: {', '.join(nominal_child)}")

print("\n" + "=" * 100)
print("✅ انتهى الطلب الثالث والرابع")
print("=" * 100)
```

```python
# ============================================================
# الطلب الخامس: توزيع المتغيرات الرقمية (Numeric Distributions)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# تعريب الخطوط
plt.rcParams['font.family'] = 'DejaVu Sans'
sns.set_style("whitegrid")

data_folder = '/content/drive/My Drive/DM Homework 2 datasets'

# تحميل البيانات
df_main = pd.read_csv(
    os.path.join(data_folder, 'womens_mens_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

df_child = pd.read_csv(
    os.path.join(data_folder, 'child_level_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

print("=" * 100)
print("الطلب الخامس: توزيع وتحليل المتغيرات الرقمية (Numeric)")
print("=" * 100)

# ============================================================
# تحديد المتغيرات الرقمية
# ============================================================

def identify_numeric_vars(df):
    """
    تحديد جميع المتغيرات الرقمية (Numeric)
    """
    numeric_vars = []
    for col in df.columns:
        if df[col].dtype in ['int64', 'int32', 'float64', 'float32']:
            numeric_vars.append(col)
    return numeric_vars

numeric_main = identify_numeric_vars(df_main)
numeric_child = identify_numeric_vars(df_child)

print(f"\n📊 المتغيرات الرقمية في womens_mens_mcm:")
print(f"   عدد المتغيرات: {len(numeric_main)}")
print(f"   أول 10: {numeric_main[:10]}")

print(f"\n📊 المتغيرات الرقمية في child_level_mcm:")
print(f"   عدد المتغيرات: {len(numeric_child)}")
print(f"   أول 10: {numeric_child[:10]}")

# ============================================================
# دالة لتحليل المتغير الرقمي
# ============================================================

def analyze_numeric_var(df, var_name, data_name="Dataset"):
    """
    تحليل شامل لمتغير رقمي واحد
    """
    data = df[var_name].dropna()

    print(f"\n{'─' * 100}")
    print(f"📈 المتغير: {var_name}")
    print(f"{'─' * 100}")

    # الإحصائيات الوصفية
    print(f"\n🔢 الإحصائيات الوصفية:")
    print(f"   • العدد (Non-Missing): {len(data)}")
    print(f"   • القيم المفقودة: {df[var_name].isna().sum()}")
    print(f"   • المتوسط (Mean): {data.mean():.4f}")
    print(f"   • الوسيط (Median): {data.median():.4f}")
    print(f"   • الانحراف المعياري (Std Dev): {data.std():.4f}")
    print(f"   • الحد الأدنى (Min): {data.min():.4f}")
    print(f"   • الحد الأقصى (Max): {data.max():.4f}")
    print(f"   • المئين 25%: {data.quantile(0.25):.4f}")
    print(f"   • المئين 75%: {data.quantile(0.75):.4f}")

    # القيم الشاذة (Outliers)
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data < lower_bound) | (data > upper_bound)]

    print(f"\n⚠️ القيم الشاذة (Outliers):")
    print(f"   • عدد القيم الشاذة: {len(outliers)}")
    print(f"   • نسبة: {(len(outliers)/len(data))*100:.2f}%")

    # التوزيع (Distribution)
    skewness = stats.skew(data)
    kurtosis = stats.kurtosis(data)

    print(f"\n📊 شكل التوزيع:")
    print(f"   • Skewness (الانحراف): {skewness:.4f}")
    if abs(skewness) < 0.5:
        print(f"     → توزيع متقارب تقريباً")
    elif skewness > 0:
        print(f"     → منحرف لليمين (Right-skewed)")
    else:
        print(f"     → منحرف لليسار (Left-skewed)")

    print(f"   • Kurtosis (التفرطح): {kurtosis:.4f}")

    # النتائج المثيرة للاهتمام
    print(f"\n💡 النتائج المثيرة للاهتمام:")

    # معامل الاختلاف (Coefficient of Variation)
    if data.mean() != 0:
        cv = (data.std() / data.mean()) * 100
        print(f"   • معامل الاختلاف (CV): {cv:.2f}%")

    # النسبة المئوية للأصفار
    zeros = (data == 0).sum()
    print(f"   • عدد الأصفار: {zeros} ({(zeros/len(data))*100:.2f}%)")

    return {
        'Mean': data.mean(),
        'Median': data.median(),
        'Std': data.std(),
        'Min': data.min(),
        'Max': data.max(),
        'Q1': Q1,
        'Q3': Q3,
        'Outliers': len(outliers),
        'Skewness': skewness,
        'N': len(data)
    }

# ============================================================
# اختيار أهم المتغيرات الرقمية للتحليل
# ============================================================

# من womens_mens - المتغيرات الأساسية والمهمة
important_numeric_main = [col for col in numeric_main
                          if col in ['id', 'wemp', 'whn', 'mhn', 'treat', 'control', 'stratum']
                          or 'age' in col.lower()
                          or 'edu' in col.lower()
                          or 'asset' in col.lower()]

if len(important_numeric_main) == 0:
    important_numeric_main = numeric_main[:5]

# من child_level - المتغيرات الأساسية
important_numeric_child = [col for col in numeric_child
                           if 'weight' in col.lower()
                           or 'height' in col.lower()
                           or 'age' in col.lower()
                           or 'sick' in col.lower()]

if len(important_numeric_child) == 0:
    important_numeric_child = numeric_child[:5]

print("\n" + "🔷" * 50)
print("البيانات: womens_mens_mcm.tab - المتغيرات الرقمية")
print("🔷" * 50)

numeric_stats_main = {}
for var in important_numeric_main[:10]:  # أول 10 متغيرات
    try:
        numeric_stats_main[var] = analyze_numeric_var(df_main, var, "womens_mens")
    except Exception as e:
        print(f"   ❌ خطأ: {e}")

print("\n" + "🔷" * 50)
print("البيانات: child_level_mcm.tab - المتغيرات الرقمية")
print("🔷" * 50)

numeric_stats_child = {}
for var in important_numeric_child[:10]:  # أول 10 متغيرات
    try:
        numeric_stats_child[var] = analyze_numeric_var(df_child, var, "child_level")
    except Exception as e:
        print(f"   ❌ خطأ: {e}")

# ============================================================
# الرسوم البيانية: Histogram و KDE Plot
# ============================================================

print("\n\n" + "=" * 100)
print("الطلب الخامس: الرسوم البيانية (Histograms)")
print("=" * 100)

def plot_numeric_var(df, var_name, data_name="Dataset"):
    """
    رسم Histogram و KDE Plot لمتغير رقمي
    """
    data = df[var_name].dropna()

    if len(data) < 2:
        return None

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Histogram
    axes[0].hist(data, bins=30, color='steelblue', alpha=0.7, edgecolor='black')
    axes[0].set_xlabel(var_name, fontsize=11, fontweight='bold')
    axes[0].set_ylabel('التكرار', fontsize=11, fontweight='bold')
    axes[0].set_title(f'Histogram: {var_name}', fontsize=12, fontweight='bold')
    axes[0].grid(axis='y', alpha=0.3)

    # KDE Plot
    axes[1].hist(data, bins=30, color='steelblue', alpha=0.5, density=True, edgecolor='black')
    try:
        data.plot(kind='kde', ax=axes[1], color='red', linewidth=2.5)
    except:
        pass
    axes[1].set_xlabel(var_name, fontsize=11, fontweight='bold')
    axes[1].set_ylabel('الكثافة', fontsize=11, fontweight='bold')
    axes[1].set_title(f'Histogram + KDE: {var_name}', fontsize=12, fontweight='bold')
    axes[1].grid(axis='y', alpha=0.3)

    fig.suptitle(f'{data_name} - {var_name}', fontsize=14, fontweight='bold')
    plt.tight_layout()

    return fig

# رسم المتغيرات الرقمية من womens_mens
print("\n📊 رسم المتغيرات الرقمية من womens_mens_mcm:")

for i, var in enumerate(important_numeric_main[:5], 1):
    try:
        print(f"   {i}. {var}")
        fig = plot_numeric_var(df_main, var, "womens_mens_mcm")
        if fig:
            plt.savefig(f'/content/drive/My Drive/WM_Numeric_Dist_{i}_{var}.png', dpi=150, bbox_inches='tight')
            plt.close()
    except Exception as e:
        print(f"      ❌ خطأ: {e}")

# رسم المتغيرات الرقمية من child_level
print("\n📊 رسم المتغيرات الرقمية من child_level_mcm:")

for i, var in enumerate(important_numeric_child[:5], 1):
    try:
        print(f"   {i}. {var}")
        fig = plot_numeric_var(df_child, var, "child_level_mcm")
        if fig:
            plt.savefig(f'/content/drive/My Drive/Child_Numeric_Dist_{i}_{var}.png', dpi=150, bbox_inches='tight')
            plt.close()
    except Exception as e:
        print(f"      ❌ خطأ: {e}")

# ============================================================
# Box Plots - الرسم الصندوقي
# ============================================================

print("\n📊 رسم Box Plots:")

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Box Plots للمتغيرات الرقمية الرئيسية', fontsize=16, fontweight='bold')

plot_idx = 0
for var in important_numeric_main[:3]:
    try:
        data = df_main[var].dropna()
        axes[0, plot_idx].boxplot(data, vert=True)
        axes[0, plot_idx].set_ylabel('القيمة', fontsize=10, fontweight='bold')
        axes[0, plot_idx].set_title(f'womens_mens: {var}', fontsize=11, fontweight='bold')
        axes[0, plot_idx].grid(axis='y', alpha=0.3)
        plot_idx += 1
    except:
        pass

plot_idx = 0
for var in important_numeric_child[:3]:
    try:
        data = df_child[var].dropna()
        axes[1, plot_idx].boxplot(data, vert=True)
        axes[1, plot_idx].set_ylabel('القيمة', fontsize=10, fontweight='bold')
        axes[1, plot_idx].set_title(f'child_level: {var}', fontsize=11, fontweight='bold')
        axes[1, plot_idx].grid(axis='y', alpha=0.3)
        plot_idx += 1
    except:
        pass

plt.tight_layout()
plt.savefig('/content/drive/My Drive/Numeric_Box_Plots.png', dpi=150, bbox_inches='tight')
plt.close()

print("✅ تم رسم Box Plots")

# ============================================================
# جداول ملخصة في Excel
# ============================================================

print("\n💾 حفظ الإحصائيات الوصفية في Excel:")

output_file = '/content/drive/My Drive/Task5_Numeric_Analysis.xlsx'

try:
    with pd.ExcelWriter(output_file, engine='openpyxl') as writer:

        # ورقة 1: ملخص الإحصائيات من womens_mens
        stats_list = []
        for var, stats_dict in numeric_stats_main.items():
            row = {'Variable': var}
            row.update(stats_dict)
            stats_list.append(row)

        if stats_list:
            stats_df_main = pd.DataFrame(stats_list)
            stats_df_main.to_excel(writer, sheet_name='WM Numeric Stats', index=False)

        # ورقة 2: ملخص الإحصائيات من child_level
        stats_list = []
        for var, stats_dict in numeric_stats_child.items():
            row = {'Variable': var}
            row.update(stats_dict)
            stats_list.append(row)

        if stats_list:
            stats_df_child = pd.DataFrame(stats_list)
            stats_df_child.to_excel(writer, sheet_name='Child Numeric Stats', index=False)

        # ورقة 3: الإحصائيات الكاملة (describe) من womens_mens
        describe_main = df_main[important_numeric_main].describe().T
        describe_main.to_excel(writer, sheet_name='WM Describe')

        # ورقة 4: الإحصائيات الكاملة من child_level
        describe_child = df_child[important_numeric_child].describe().T
        describe_child.to_excel(writer, sheet_name='Child Describe')

    print(f"✅ تم الحفظ: {output_file}")

except Exception as e:
    print(f"⚠️ خطأ: {e}")

# ============================================================
# ملخص النتائج
# ============================================================

print("\n" + "=" * 100)
print("📌 ملخص الطلب الخامس")
print("=" * 100)

print("\n🔍 womens_mens_mcm.tab:")
print(f"   • عدد المتغيرات الرقمية: {len(numeric_main)}")
print(f"   • المتغيرات المحللة: {len(numeric_stats_main)}")

print("\n🔍 child_level_mcm.tab:")
print(f"   • عدد المتغيرات الرقمية: {len(numeric_child)}")
print(f"   • المتغيرات المحللة: {len(numeric_stats_child)}")

print("\n✅ انتهى الطلب الخامس")
print("=" * 100)
```

```python
# ============================================================
# الطلب السادس: Box Plot لأوزان الأطفال (مُصحح - الإصدار النهائي)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import os
import warnings
warnings.filterwarnings('ignore')

# تعريب الخطوط
plt.rcParams['font.family'] = 'DejaVu Sans'
sns.set_style("whitegrid")

data_folder = '/content/drive/My Drive/DM Homework 2 datasets'

# تحميل البيانات
print("📥 تحميل البيانات...")

df_main = pd.read_csv(
    os.path.join(data_folder, 'womens_mens_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

df_child = pd.read_csv(
    os.path.join(data_folder, 'child_level_mcm.tab'),
    sep='\t',
    encoding='utf-8',
    low_memory=False
)

print(f"✅ تم تحميل البيانات:")
print(f"   • womens_mens_mcm: {len(df_main)} صف")
print(f"   • child_level_mcm: {len(df_child)} صف")

print("\n" + "=" * 100)
print("الطلب السادس: Box Plot لأوزان الأطفال - قبل وبعد الدراسة")
print("=" * 100)

# ============================================================
# تنظيف البيانات المدمجة بالفعل
# ============================================================

baseline_var = 'c_weight'
endline_var = 'Ec_weight'

print(f"\n\n📊 تنظيف بيانات الوزن...")

# df_child بالفعل يحتوي على treat و control
# نتحقق من ذلك
print(f"\n✅ تم التحقق:")
print(f"   • treat في df_child: {'نعم' if 'treat' in df_child.columns else 'لا'}")
print(f"   • control في df_child: {'نعم' if 'control' in df_child.columns else 'لا'}")

# استخدام df_child مباشرة (بدون دمج إضافي)
df_analysis = df_child[[baseline_var, endline_var, 'treat', 'control']].copy()

# حذف الصفوف ذات القيم المفقودة في Baseline
df_analysis = df_analysis[df_analysis[baseline_var].notna()]
print(f"✅ بعد حذف القيم المفقودة (Baseline): {len(df_analysis)} صف")

# حذف القيم غير المعقولة (أقل من 1 كغ أو أكثر من 30 كغ)
df_analysis = df_analysis[(df_analysis[baseline_var] > 1) & (df_analysis[baseline_var] < 30)]
print(f"✅ بعد حذف القيم الشاذة: {len(df_analysis)} صف")

print(f"\n📊 إحصائيات Baseline Weight:")
print(f"   • المتوسط: {df_analysis[baseline_var].mean():.2f} كغ")
print(f"   • الوسيط: {df_analysis[baseline_var].median():.2f} كغ")
print(f"   • الحد الأدنى: {df_analysis[baseline_var].min():.2f} كغ")
print(f"   • الحد الأقصى: {df_analysis[baseline_var].max():.2f} كغ")
print(f"   • الانحراف المعياري: {df_analysis[baseline_var].std():.2f} كغ")

# ============================================================
# فحص توزيع المجموعات
# ============================================================

print(f"\n\n📋 توزيع المجموعات في البيانات:")

for group in [1, 2, 3, 4]:
    count = (df_analysis['treat'] == group).sum()
    if count > 0:
        print(f"   • المجموعة {group}: {count} طفل")

control_count = (df_analysis['control'] == 1).sum()
treatment_count = (df_analysis['control'] == 0).sum()
print(f"\n   • مجموعة التحكم (control=1): {control_count} طفل")
print(f"   • مجموعة العلاج (control=0): {treatment_count} طفل")

# ============================================================
# رسم Box Plot - المجموعات الأربع
# ============================================================

print("\n\n" + "=" * 100)
print("📊 رسم Box Plot - المجموعات الأربع")
print("=" * 100)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('توزيع أوزان الأطفال حسب مجموعات العلاج\n(Baseline Weight - c_weight)',
             fontsize=16, fontweight='bold', y=0.995)

colors = {1: '#FF6B6B', 2: '#4ECDC4', 3: '#45B7D1', 4: '#FFA07A'}
group_labels = {
    1: 'المجموعة 1',
    2: 'المجموعة 2',
    3: 'المجموعة 3',
    4: 'المجموعة 4'
}

plot_count = 0
for idx, treat_group in enumerate([1, 2, 3, 4], 1):
    ax = axes[(idx-1)//2, (idx-1)%2]

    # تصفية البيانات
    group_data = df_analysis[df_analysis['treat'] == treat_group][baseline_var].dropna()

    if len(group_data) > 0:
        plot_count += 1
        # رسم Box Plot
        bp = ax.boxplot(group_data, labels=[f'{group_labels[treat_group]}'],
                        patch_artist=True, widths=0.6, vert=True)

        # تلوين
        for patch in bp['boxes']:
            patch.set_facecolor(colors[treat_group])
            patch.set_alpha(0.7)

        # الإحصائيات
        count = len(group_data)
        mean_val = group_data.mean()
        median_val = group_data.median()
        std_val = group_data.std()
        q1 = group_data.quantile(0.25)
        q3 = group_data.quantile(0.75)

        stats_text = (f"ن = {count}\n"
                     f"المتوسط = {mean_val:.2f}\n"
                     f"الوسيط = {median_val:.2f}\n"
                     f"الانحراف = {std_val:.2f}\n"
                     f"Q1 = {q1:.2f}, Q3 = {q3:.2f}")

        ax.text(0.98, 0.97, stats_text, transform=ax.transAxes,
               fontsize=9, verticalalignment='top', horizontalalignment='right',
               bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.7),
               fontfamily='monospace')

        ax.set_ylabel('الوزن (كغ)', fontsize=11, fontweight='bold')
        ax.set_title(f'{group_labels[treat_group]} (ن={count})', fontsize=12, fontweight='bold')
        ax.grid(axis='y', alpha=0.3)
    else:
        ax.text(0.5, 0.5, 'لا توجد بيانات', transform=ax.transAxes,
               ha='center', va='center', fontsize=14)
        ax.set_title(f'{group_labels[treat_group]} (بدون بيانات)', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.savefig('/content/drive/My Drive/Task6_BoxPlot_ByTreatment.png', dpi=150, bbox_inches='tight')
plt.close()

print(f"✅ تم رسم Box Plot حسب مجموعات العلاج ({plot_count} مجموعات)")

# ============================================================
# رسم Box Plot - Control vs Treatment
# ============================================================

print("\n📊 رسم Box Plot - مجموعة التحكم vs مجموعة العلاج:")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('مقارنة أوزان الأطفال: مجموعة التحكم vs مجموعة العلاج\n(Baseline Weight - c_weight)',
             fontsize=14, fontweight='bold')

# استخراج البيانات
control_data = df_analysis[df_analysis['control'] == 1][baseline_var].dropna()
treatment_data = df_analysis[df_analysis['control'] == 0][baseline_var].dropna()

print(f"   • مجموعة التحكم: {len(control_data)} طفل")
print(f"   • مجموعة العلاج: {len(treatment_data)} طفل")

# رسم Box Plot
data_to_plot = [control_data, treatment_data]
labels = ['مجموعة التحكم', 'مجموعة العلاج']

bp = axes[0].boxplot(data_to_plot, labels=labels, patch_artist=True, widths=0.6)

for patch, color in zip(bp['boxes'], ['#90EE90', '#FFB6C6']):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

axes[0].set_ylabel('الوزن (كغ)', fontsize=11, fontweight='bold')
axes[0].set_title('Box Plot - المقارنة', fontsize=12, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)

# الإحصائيات في الجانب الأيمن
stats_comparison = f"""مجموعة التحكم (Control):
  • العدد: {len(control_data)}
  • المتوسط: {control_data.mean():.2f} كغ
  • الوسيط: {control_data.median():.2f} كغ
  • الانحراف: {control_data.std():.2f} كغ
  • النطاق: {control_data.min():.2f} - {control_data.max():.2f}

مجموعة العلاج (Treatment):
  • العدد: {len(treatment_data)}
  • المتوسط: {treatment_data.mean():.2f} كغ
  • الوسيط: {treatment_data.median():.2f} كغ
  • الانحراف: {treatment_data.std():.2f} كغ
  • النطاق: {treatment_data.min():.2f} - {treatment_data.max():.2f}"""

axes[1].text(0.05, 0.95, stats_comparison, transform=axes[1].transAxes,
            fontsize=9, verticalalignment='top', fontfamily='monospace',
            bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7))
axes[1].axis('off')

plt.tight_layout()
plt.savefig('/content/drive/My Drive/Task6_BoxPlot_Control_vs_Treatment.png', dpi=150, bbox_inches='tight')
plt.close()

print("✅ تم رسم مقارنة Control vs Treatment")

# ============================================================
# الاختبارات الإحصائية
# ============================================================

print("\n\n" + "=" * 100)
print("🔬 الاختبارات الإحصائية")
print("=" * 100)

# T-test
if len(control_data) > 1 and len(treatment_data) > 1:
    t_stat, p_value = stats.ttest_ind(control_data, treatment_data)

    print(f"\n📊 Independent t-test (Control vs Treatment):")
    print(f"   • t-statistic: {t_stat:.4f}")
    print(f"   • p-value: {p_value:.6f}")
    print(f"   • Significant (α=0.05): {'✅ نعم (معنوي إحصائياً)' if p_value < 0.05 else '❌ لا (غير معنوي)'}")

    # حساب Cohen's d
    pooled_std = np.sqrt(((len(control_data)-1)*control_data.std()**2 +
                           (len(treatment_data)-1)*treatment_data.std()**2) /
                          (len(control_data) + len(treatment_data) - 2))

    if pooled_std > 0:
        cohens_d = (control_data.mean() - treatment_data.mean()) / pooled_std
        print(f"   • Cohen's d (حجم التأثير): {cohens_d:.4f}")

        if abs(cohens_d) < 0.2:
            effect_size = "تأثير صغير جداً"
        elif abs(cohens_d) < 0.5:
            effect_size = "تأثير صغير"
        elif abs(cohens_d) < 0.8:
            effect_size = "تأثير متوسط"
        else:
            effect_size = "تأثير كبير"
        print(f"     → {effect_size}")

# ANOVA بين المجموعات الأربع
print(f"\n" + "─" * 100)

groups_data = []
for i in [1, 2, 3, 4]:
    g = df_analysis[df_analysis['treat'] == i][baseline_var].dropna()
    if len(g) > 0:
        groups_data.append(g)

if len(groups_data) > 1:
    f_stat, p_value_anova = stats.f_oneway(*groups_data)

    print(f"\n📊 ANOVA Test (بين المجموعات الأربع):")
    print(f"   • F-statistic: {f_stat:.4f}")
    print(f"   • p-value: {p_value_anova:.6f}")
    print(f"   • Significant (α=0.05): {'✅ نعم (معنوي إحصائياً)' if p_value_anova < 0.05 else '❌ لا (غير معنوي)'}")

# ============================================================
# جدول ملخص
# ============================================================

print("\n\n" + "=" * 100)
print("📋 ملخص الإحصائيات حسب المجموعات")
print("=" * 100)

summary_data = []
for group in [1, 2, 3, 4]:
    group_data = df_analysis[df_analysis['treat'] == group][baseline_var].dropna()
    if len(group_data) > 0:
        summary_data.append({
            'المجموعة': f'مجموعة {group}',
            'العدد': len(group_data),
            'المتوسط': f"{group_data.mean():.2f}",
            'الوسيط': f"{group_data.median():.2f}",
            'الانحراف': f"{group_data.std():.2f}",
            'Min': f"{group_data.min():.2f}",
            'Max': f"{group_data.max():.2f}"
        })

summary_df = pd.DataFrame(summary_data)
print("\n" + summary_df.to_string(index=False))

# ============================================================
# حفظ النتائج
# ============================================================

print("\n\n💾 حفظ النتائج في Excel:")

output_file = '/content/drive/My Drive/Task6_BoxPlot_Analysis.xlsx'

try:
    with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
        # ورقة 1: الملخص
        summary_df.to_excel(writer, sheet_name='Summary Statistics', index=False)

        # ورقة 2: البيانات الخام
        df_analysis.to_excel(writer, sheet_name='Raw Data', index=False)

        # ورقة 3: النتائج الإحصائية
        if len(control_data) > 1 and len(treatment_data) > 1:
            stats_results = {
                'الاختبار': ['t-test', 'ANOVA'],
                'Test Statistic': [f"{t_stat:.4f}", f"{f_stat:.4f}"],
                'P-value': [f"{p_value:.6f}", f"{p_value_anova:.6f}"],
                'معنوي إحصائياً': ['نعم' if p_value < 0.05 else 'لا',
                                'نعم' if p_value_anova < 0.05 else 'لا'],
                'ملاحظات': [f"Cohen's d = {cohens_d:.4f}", 'ANOVA - 4 Groups']
            }
            stats_df = pd.DataFrame(stats_results)
            stats_df.to_excel(writer, sheet_name='Statistical Tests', index=False)

    print(f"✅ تم الحفظ: {output_file}")

except Exception as e:
    print(f"⚠️ خطأ: {e}")

# ============================================================
# الاستنتاجات النهائية
# ============================================================

print("\n\n" + "=" * 100)
print("💡 الاستنتاجات الرئيسية")
print("=" * 100)

print(f"""
1️⃣ توزيع أوزان الأطفال (Baseline - c_weight):
   • المتوسط العام: {df_analysis[baseline_var].mean():.2f} كغ
   • الوسيط العام: {df_analysis[baseline_var].median():.2f} كغ
   • النطاق: {df_analysis[baseline_var].min():.2f} - {df_analysis[baseline_var].max():.2f} كغ
   • الانحراف المعياري: {df_analysis[baseline_var].std():.2f} كغ

2️⃣ المقارنة بين مجموعة التحكم والعلاج:
   • مجموعة التحكم:
     - العدد = {len(control_data)}
     - المتوسط = {control_data.mean():.2f} كغ
   • مجموعة العلاج:
     - العدد = {len(treatment_data)}
     - المتوسط = {treatment_data.mean():.2f} كغ
   • الفرق = {abs(control_data.mean() - treatment_data.mean()):.2f} كغ

3️⃣ النتائج الإحصائية:
   • t-test p-value: {p_value:.6f} {'✅ (معنوي إحصائياً)' if p_value < 0.05 else '❌ (غير معنوي)'}
   • ANOVA p-value: {p_value_anova:.6f} {'✅ (معنوي إحصائياً)' if p_value_anova < 0.05 else '❌ (غير معنوي)'}
   • حجم التأثير (Cohen's d): {cohens_d:.4f} - {effect_size}

4️⃣ الملاحظات الهامة:
   ✓ البيانات تحتوي على أوزان الأطفال بشكل معقول
   ✓ توجد بيانات Endline Weight (Ec_weight) متاحة للمقارنة قبل/بعد
   ✓ العينة كافية للتحليل الإحصائي
   ✓ البيانات موزعة بشكل معقول عبر المجموعات
""")

print("\n" + "=" * 100)
print("✅ انتهى الطلب السادس بنجاح")
print("=" * 100)
```
